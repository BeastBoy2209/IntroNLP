## 1. Загрузка и подготовка данных

```python
from src.download_data import download_jigsaw_dataset, download_spam_dataset
from src.dataset import prepare_multiclass_dataset

# Загрузка исходных данных
download_jigsaw_dataset()
download_spam_dataset()

# Формирование train/test (80/20)
train_df, test_df = prepare_multiclass_dataset()
```

## 2. Разведочный анализ данных (EDA)

```python
print(f"Размер обучающей выборки (train): {train_df.shape}")
print(f"Размер тестовой выборки (test): {test_df.shape}")
train_df.head(10)
```

### Распределение классов

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=train_df, x='label', palette='viridis')
plt.title('Распределение классов (Обучающая выборка)')
plt.xlabel('Класс')
plt.ylabel('Количество')
plt.show()

print(train_df['label'].value_counts(normalize=True).mul(100).round(2).astype(str) + '%')
```

### Длина сообщений (символы и слова)

```python
train_df['char_length'] = train_df['text'].astype(str).apply(len)
train_df['word_count'] = train_df['text'].astype(str).apply(lambda x: len(x.split()))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(ax=axes[0], data=train_df, x='label', y='char_length', showfliers=False, palette='Set2')
axes[0].set_title('Длина в символах (без выбросов)')

sns.boxplot(ax=axes[1], data=train_df, x='label', y='word_count', showfliers=False, palette='Pastel1')
axes[1].set_title('Количество слов (без выбросов)')
plt.tight_layout()
plt.show()

train_df.groupby('label')[['char_length', 'word_count']].describe()
```
